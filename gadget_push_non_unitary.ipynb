# Gadget Push Non Unitary Playground

```python
from faultgadget.graph_helpers import add_sinks_for_all_detecting_regions, add_all_gadgets
import pyzx as zx
from faultgadget import GadgetGraph, generate, compute_webs_for_gadgets
from faultgadget.web import compute_detecting_regions

og = generate.zweb(2, 2)
g = GadgetGraph.from_graph(og)
zx.draw(g, labels=True)

add_sinks_for_all_detecting_regions(g)
gc, _ = g.realise()
zx.draw(gc, labels=True)
add_all_gadgets(g)
gc, _ = g.realise()
zx.draw(gc, labels=True)
```

```python
detecting_regions = compute_detecting_regions(og)
for region in detecting_regions: zx.draw(og, pauli_web=region, labels=True)
```

```python
webs = compute_webs_for_gadgets(g, g.gadgets().keys())
for gadget_id in g.gadgets(): zx.draw(gc, pauli_web=webs[gadget_id], labels=True)
```
