# Gadget Push Demo

```python
import random

random.seed(50)

from faultgadget import GadgetGraph
from pyzx.editor_actions import match_hadamard_edge
from pyzx.hrules import had_edge_to_hbox

import pyzx as zx

g = zx.generate.cliffords(9, 9)
zx.clifford_simp(g, quiet=True)
for e in match_hadamard_edge(g):
    had_edge_to_hbox(g, e)

dg = GadgetGraph.from_graph(g)
dg.add_all_gadgets()

g, _ = dg.realise()
g.pack_circuit_rows()
g.pack_circuit_qubits()
```

```python
from faultgadget import compute_webs_for_gadgets

webs = compute_webs_for_gadgets(dg, dg.gadgets().keys())
if len(webs) == 0: print("No webs!")
for _, web in webs.items(): zx.draw(g, pauli_web=web, labels=True)
```

```python
from faultgadget import expand_all_gadgets
from pyzx.hsimplify import from_hypergraph_form

expand_all_gadgets(dg)

g = dg.realise()
from_hypergraph_form(g)
zx.full_reduce(g)
zx.tensorfy(g, preserve_scalar=False)
```
