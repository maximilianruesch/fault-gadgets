# Gadget Push Demo

```python
import random

random.seed(50)

import pyzx as zx

g = zx.generate.cnots(4, 5)
zx.id_simp(g)
```

```python
from faultgadget import GadgetGraph

bg = GadgetGraph.from_graph(g)
_, _, y_gadget_id = bg.add_edge_flip_gadgets((9, 12))

bgc, _ = bg.realise()
zx.draw(bgc, labels=True)
```

```python
from faultgadget import compute_web_for_gadget, fire_web_onto_gadget

debug = dict()
webs = None
try:
    web = compute_web_for_gadget(graph=bg, gadget_id=y_gadget_id)
except Exception as e:
    debug['g'].pack_circuit_rows()
    for i, web in enumerate(debug['g_webs']):
        zx.draw(debug['g'], pauli_web=web, labels=True)
    raise e
    
debug['g'].pack_circuit_rows()
bgc.pack_circuit_rows()

bgg = bg.clone(GadgetGraph())

print("Firing web mapped onto original graph:")
zx.draw(bgc, pauli_web=web, labels=True)
fire_web_onto_gadget(g=bgg, gadget_id=y_gadget_id, web=web)
bgc1 = bgg.clone(GadgetGraph())
bgc1.full_instance()
print("Graph with fired web for gadget:")
zx.draw(bgc1, labels=True)
bgg.merge_targets()
bgc2 = bgg.clone(GadgetGraph())
bgc2.full_instance()
print("Merged gadget:")
zx.draw(bgc2, labels=True)
print("Firing finished!")
```

```python
from faultgadget import expand_all_gadgets

bg = GadgetGraph.from_graph(g)
bg.add_all_gadgets()
bg.reassign_gadget_positions()
bg.pack_circuit_rows()
zx.draw(bg, labels=True)
expand_all_gadgets(bg)
bg.full_instance()
zx.draw(bg, labels=True)
```
