# Computing Pauli Webs

```python
import random

random.seed(50)

from faultgadget.web import compute_detecting_regions, compute_stabilisers

import pyzx as zx

g = zx.generate.cliffords(9, 9)
```

```python
stabilisers = compute_stabilisers(g)
if len(stabilisers) == 0: print("No stabilisers :(")
for stabiliser in stabilisers: zx.draw(g, pauli_web=stabiliser)
```

```python
regions = compute_detecting_regions(g)
if len(regions) == 0: print("No detecting regions :(")
for region in regions: zx.draw(g, pauli_web=region)
```
