# Sampling fault signatures:

```python
import time
import pyzx as zx
from faultgadget.query.sampling import sample_signatures

g = zx.generate.cnots(3, 4, backend='simple')
start_time = time.time()
for i, sig in enumerate(sample_signatures(g, shots=10000000)):
    if i == 0:
        print(f"Startup time: {time.time() - start_time} seconds!")
        start_time = time.time()

    if i % 100000 == 0:
        print(f"Sampling at {i / (time.time() - start_time)} Hz...")
```
