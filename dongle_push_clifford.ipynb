# Dongle Push Demo

```python
import random

random.seed(50)

import pyzx as zx
from pyzx.editor_actions import match_hadamard_edge
from pyzx.hrules import had_edge_to_hbox

g = zx.generate.cliffords(4, 7)
zx.clifford_simp(g, quiet=True)
g.normalize()
for e in match_hadamard_edge(g):
    had_edge_to_hbox(g, e)
zx.draw(g, labels=True)
```

```python
from dongle import DongleGraph, expand_all_dongles

bg = DongleGraph.from_graph(g)
bg.add_all_dongles()
bg.reassign_dongle_positions()
bg.pack_circuit_rows()
bg.pack_circuit_qubits()
zx.draw(bg, labels=True)
expand_all_dongles(bg)
bg.full_instance()
zx.draw(bg, labels=True)
```
