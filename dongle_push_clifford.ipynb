# Dongle Push Demo

```python
import random

random.seed(50)

from dongle import DongleGraph
from pyzx.editor_actions import match_hadamard_edge
from pyzx.hrules import had_edge_to_hbox

import pyzx as zx

g = zx.generate.cliffords(9, 9)
zx.clifford_simp(g, quiet=True)
for e in match_hadamard_edge(g):
    had_edge_to_hbox(g, e)

dg = DongleGraph.from_graph(g)
dg.add_all_dongles()

g = dg.clone(DongleGraph())
g.reassign_dongle_positions()
g.pack_circuit_rows()
g.pack_circuit_qubits()
g.realise_all_targets()
```

```python
from dongle import compute_webs_for_dongles

webs = compute_webs_for_dongles(dg, dg.dongles().keys())
if len(webs) == 0: print("No webs!")
for dongle_id, web in webs.items(): zx.draw(g, pauli_web=web, labels=True)
```

```python
from dongle import expand_all_dongles
from pyzx.hsimplify import from_hypergraph_form

expand_all_dongles(dg)

g = dg.clone(DongleGraph())
g.realise_all_targets()
g.reassign_dongle_positions()
from_hypergraph_form(g)
zx.full_reduce(g)
zx.tensorfy(g, preserve_scalar=False)
```
