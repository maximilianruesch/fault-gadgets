# Dongle Push Non Unitary Playground

```python
import pyzx as zx
from dongle import DongleGraph, generate

og = generate.zweb(5, 5)
g = DongleGraph.from_graph(og)
zx.draw(g, labels=True)

g.add_all_dongles()
gc = g.clone(DongleGraph())
gc.reassign_dongle_positions()
gc.pack_circuit_rows()
gc.pack_circuit_qubits()
gc.realise_all_targets()
zx.draw(gc, labels=True)
```

```python
from dongle.web import compute_detecting_regions

detecting_regions = compute_detecting_regions(og)
for region in detecting_regions: zx.draw(og, pauli_web=region, labels=True)
```

```python
from dongle import compute_webs_for_dongles

webs = compute_webs_for_dongles(g, g.dongles().keys())
for dongle_id in g.dongles(): zx.draw(gc, pauli_web=webs[dongle_id], labels=True)
```
